```python
import pandas as pd

# ---------------------------------------------------------------
# LOAD DATASETS DIRECTLY FROM GITHUB
# Dataset source: github.com/getdebugged/northstar_case_study
# ---------------------------------------------------------

base_url = 'https://raw.githubusercontent.com/getdebugged/case_study/main/'

app_events  = pd.read_csv(base_url + 'app_events.csv')
complaints  = pd.read_csv(base_url + 'complaints.csv')
customers   = pd.read_csv(base_url + 'customers.csv')
deliveries  = pd.read_csv(base_url + 'deliveries.csv')
drivers     = pd.read_csv(base_url + 'drivers.csv')
hubs        = pd.read_csv(base_url + 'hubs.csv')
incidents   = pd.read_csv(base_url + 'incidents.csv')
orders      = pd.read_csv(base_url + 'orders.csv')
vehicles    = pd.read_csv(base_url + 'vehicles.csv')

# Collect all DataFrames into a dictionary to easily iterate over them
dataframes = {
    'app_events': app_events,
    'complaints': complaints,
    'customers': customers,
    'deliveries': deliveries,
    'drivers': drivers,
    'hubs': hubs,
    'incidents': incidents,
    'orders': orders,
    'vehicles': vehicles
}

print(' All datasets loaded directly from GitHub!')
print(f'Orders: {len(orders)} rows')
print(f'Deliveries: {len(deliveries)} rows')
print(f'Complaints: {len(complaints)} rows')
```

```python
# Preview each dataset - see columns and first 3 rows

for name, df in dataframes.items():
    print(f'\n{"="*50}')
    print(f' {name.upper()}')
    print(f'Rows: {df.shape[0]} | Columns: {df.shape[1]}')
    print(f'{"="*50}')
    display(df.head(3))
```

## 1. Python Data Analysis
### Data Cleaning and Exploration


```python
# Check for missing values in all datasets
print("=== MISSING VALUES ===\n")
for name, df in dataframes.items():
    missing = df.isnull().sum().sum()
    if missing > 0:
        print(f'{name}: {missing} missing values')
        print(df.isnull().sum())
        print()
    else:
        print(f' {name}: No missing values')
```

## Data Cleaning

```python
#---------------------------------------------
# DATA CLEANING
# --------------------------------------------

# 1. app_events - order_id is missing because some events
#    are not linked to orders (e.g. browsing) - fill with 'No Order'
app_events['order_id'] = app_events['order_id'].fillna('No Order')

# 2. customers - fill missing loyalty_score with average score
#    fill missing preferred_channel with 'Unknown'
customers['loyalty_score'] = customers['loyalty_score'].fillna(customers['loyalty_score'].mean())
customers['preferred_channel'] = customers['preferred_channel'].fillna('Unknown')

# 3. complaints - fill missing compensation_amount with 0
#    (no compensation was given)
complaints['compensation_amount'] = complaints['compensation_amount'].fillna(0)

# 4. deliveries - missing delivery_completed_at means not completed
#    fill missing customer_rating with 0
deliveries['delivery_completed_at'] = deliveries['delivery_completed_at'].fillna('Not Completed')
deliveries['customer_rating_post_delivery'] = deliveries['customer_rating_post_delivery'].fillna(0)

# 5. drivers - fill missing training_score with average
drivers['training_score'] = drivers['training_score'].fillna(drivers['training_score'].mean())

# 6. vehicles - fill missing battery_health with average
vehicles['battery_health_pct'] = vehicles['battery_health_pct'].fillna(vehicles['battery_health_pct'].mean())

# 7. incidents - fill missing resolved_hours with average
incidents['resolved_hours'] = incidents['resolved_hours'].fillna(incidents['resolved_hours'].mean())

# 8. orders - fill missing booking_channel with 'Unknown'
orders['booking_channel'] = orders['booking_channel'].fillna('Unknown')

# Verify all cleaned
print("=== AFTER CLEANING ===\n")
for name, df in dataframes.items():
    missing = df.isnull().sum().sum()
    if missing > 0:
        print(f' {name}: still has {missing} missing values')
    else:
        print(f' {name}: Clean!')
```

## 2. Delivery Performance Analysis

```python
import matplotlib.pyplot as plt

# ---------------------------------------------
# DELIVERY PERFORMANCE ANALYSIS
# How many deliveries succeeded vs failed?
# ---------------------------------------------

# Count each delivery status
status_counts = deliveries['delivery_status'].value_counts()

# Create a bar chart
plt.figure(figsize=(8,5))
plt.bar(status_counts.index, status_counts.values, color=['green','red','orange'])
plt.title('Delivery Status Overview')
plt.xlabel('Delivery Status')
plt.ylabel('Number of Deliveries')
plt.tight_layout()
plt.show()

# Print the numbers too
print(status_counts)
print(f'\nTotal deliveries: {len(deliveries)}')
print(f'Failure rate: {round(status_counts.get("Failed",0)/len(deliveries)*100, 1)}%')
```

35% of deliveries had problems — 202 delayed and 132 failed out of 950 total.

### Delivery Failures by Hub

```python
# -------------------------------------------
# WHICH HUB HAS THE MOST FAILURES?
# join deliveries with hubs to see
# which hub is causing the most problems
# -------------------------------------------

# Merge deliveries and hubs tables together
# This is like a JOIN in SQL - connecting two tables
delivery_hub = deliveries.merge(hubs, on='hub_id', how='left')

# Filter only failed deliveries
failed = delivery_hub[delivery_hub['delivery_status'] == 'Failed']

# Count failures per hub
failures_by_hub = failed['hub_name'].value_counts()

# Draw the chart
plt.figure(figsize=(8,5))
plt.bar(failures_by_hub.index, failures_by_hub.values, color='red')
plt.title('Delivery Failures by Hub')
plt.xlabel('Hub Name')
plt.ylabel('Number of Failed Deliveries')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

print(failures_by_hub)
```

Midtown Relay and Central Core are the worst performing hubs with 26 and 23 failures respectively. These two hubs need urgent investigation.

### Manual Route Overrides Analysis

```python
# ---------------------------------------------
# MANUAL ROUTE OVERRIDES
# The case study mentions drivers are doing
# manual route overrides - let's investigate
# ---------------------------------------------

# Average manual overrides per hub
overrides_by_hub = delivery_hub.groupby('hub_name')['manual_route_override_count'].mean().round(2)
overrides_by_hub = overrides_by_hub.sort_values(ascending=False)

plt.figure(figsize=(8,5))
plt.bar(overrides_by_hub.index, overrides_by_hub.values, color='purple')
plt.title('Average Manual Route Overrides by Hub')
plt.xlabel('Hub Name')
plt.ylabel('Average Overrides per Delivery')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

print(overrides_by_hub)
```

Midtown Relay has the highest override rate (1.11 per delivery) — this links directly to why it also has the most failures. Drivers are constantly changing routes which causes delays.

### Customer Complaints Analysis

```python
# ---------------------------------------------
# CUSTOMER COMPLAINTS ANALYSIS
# What are customers complaining about most?
# --------------------------------------------

# Count complaint types
complaint_types = complaints['complaint_type'].value_counts()

plt.figure(figsize=(8,5))
plt.bar(complaint_types.index, complaint_types.values, color='darkred')
plt.title('Customer Complaints by Type')
plt.xlabel('Complaint Type')
plt.ylabel('Number of Complaints')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

print(complaint_types)

# Complaints by severity
print('\n--- Complaints by Severity ---')
print(complaints['severity'].value_counts())
```

Delay is the #1 complaint with 101 cases, followed by MissedPickup (64). Together they make up 51% of all complaints — directly reflecting the delivery problems we saw earlier.

### Complaint Resolution Analysis

```python
# -----------------------------------------------------
# HOW LONG DOES IT TAKE TO RESOLVE COMPLAINTS?
# High resolution days = poor customer service
# ----------------------------------------------------

# Average resolution days per complaint type
resolution = complaints.groupby('complaint_type')['resolution_days'].mean().round(1)
resolution = resolution.sort_values(ascending=False)

plt.figure(figsize=(8,5))
plt.bar(resolution.index, resolution.values, color='orange')
plt.title('Average Complaint Resolution Days by Type')
plt.xlabel('Complaint Type')
plt.ylabel('Average Days to Resolve')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

print(resolution)
```

Damage complaints take the longest to resolve (11.3 days) which is very poor customer service. Even Delay complaints take 7.3 days to resolve.

### Driver Performance Analysis

```python
# ----------------------------------------------
# DRIVER PERFORMANCE ANALYSIS
# Are experienced drivers performing better?
# ----------------------------------------------

# Join deliveries with drivers
delivery_driver = deliveries.merge(drivers, on='driver_id', how='left')

# Average delivery rating by employment type
rating_by_employment = delivery_driver.groupby('employment_type')['customer_rating_post_delivery'].mean().round(2)

plt.figure(figsize=(6,4))
plt.bar(rating_by_employment.index, rating_by_employment.values, color='steelblue')
plt.title('Average Customer Rating by Driver Employment Type')
plt.xlabel('Employment Type')
plt.ylabel('Average Rating')
plt.tight_layout()
plt.show()

print(rating_by_employment)

# Failed deliveries by employment type
print('\n--- Failed Deliveries by Employment Type ---')
failed_by_emp = delivery_driver[delivery_driver['delivery_status']=='Failed']['employment_type'].value_counts()
print(failed_by_emp)
```

Part-time drivers actually have the highest rating (3.86) but full-time drivers have the most failures (85). This is because there are simply more full-time drivers, not because they perform worse.



### Vehicle and Incident Analysis

```python
# -----------------------------------------------
# VEHICLE INCIDENTS ANALYSIS
# Which incident types take longest to resolve?
# What vehicle types have most incidents?
# -----------------------------------------------

# Join incidents with deliveries to get vehicle info
incident_vehicle = incidents.merge(deliveries[['delivery_id','vehicle_id']], on='delivery_id', how='left')
incident_vehicle = incident_vehicle.merge(vehicles[['vehicle_id','vehicle_type']], on='vehicle_id', how='left')

# Incidents by type
incident_counts = incidents['incident_type'].value_counts()

plt.figure(figsize=(8,5))
plt.bar(incident_counts.index, incident_counts.values, color='brown')
plt.title('Incidents by Type')
plt.xlabel('Incident Type')
plt.ylabel('Number of Incidents')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

print(incident_counts)

# Average resolution hours by incident type
print('\n--- Average Resolution Hours by Incident Type ---')
res_hours = incidents.groupby('incident_type')['resolved_hours'].mean().round(1).sort_values(ascending=False)
print(res_hours)
```

ProofMissing (46) and CustomerNoShow (44) are most common. But CustomerNoShow takes the longest to resolve at 13.8 hours — meaning when customers don't show up, it causes long operational delays.

### Order Performance and Cost Analysis

```python
# -----------------------------------------------
# ORDER PERFORMANCE ANALYSIS
# Which service types have most problems?
# ------------------------------------------------

# Orders by service type
service_counts = orders['service_type'].value_counts()

plt.figure(figsize=(7,4))
plt.bar(service_counts.index, service_counts.values, color='teal')
plt.title('Orders by Service Type')
plt.xlabel('Service Type')
plt.ylabel('Number of Orders')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# Average order value by service type
avg_value = orders.groupby('service_type')['order_value'].mean().round(2).sort_values(ascending=False)
print('--- Average Order Value by Service Type ---')
print(avg_value)

# Priority level distribution
print('\n--- Orders by Priority Level ---')
print(orders['priority_level'].value_counts())
```

Passenger (340) is the most common service type but has the highest average order value (£96.07). There are 91 Critical priority orders which need special attention.

### Fuel Cost vs Delivery Status

```python
# ----------------------------------------------
# FUEL/CHARGE COST ANALYSIS
# Do failed deliveries cost more in fuel?
# This shows financial impact of failures
# ----------------------------------------------

# Average fuel cost by delivery status
fuel_by_status = deliveries.groupby('delivery_status')['fuel_or_charge_cost'].mean().round(2)

plt.figure(figsize=(6,4))
plt.bar(fuel_by_status.index, fuel_by_status.values, color=['green','orange','red'])
plt.title('Average Fuel/Charge Cost by Delivery Status')
plt.xlabel('Delivery Status')
plt.ylabel('Average Cost (£)')
plt.tight_layout()
plt.show()

print(fuel_by_status)

# Total cost wasted on failed deliveries
total_failed_cost = deliveries[deliveries['delivery_status']=='Failed']['fuel_or_charge_cost'].sum()
print(f'\n Total fuel cost wasted on failed deliveries: £{round(total_failed_cost,2)}')
```

Failed and Delayed deliveries cost slightly MORE in fuel (£13.15, £13.14) than successful ones (£12.68). Total fuel wasted on failed deliveries alone is £1,735.53 — a direct financial loss to NorthStar.
